```python
# Importacao das bibliotecas
import pandas as pd
import os
```

```python
# Camada Bronze
# Pastas de dados: dados/ (brutos) e dados/bronze/ (saida desta etapa)
PASTA_DADOS = os.path.join("..", "dados")
PASTA_BRONZE = os.path.join(PASTA_DADOS, "bronze")
os.makedirs(PASTA_BRONZE, exist_ok=True)
```

```python
# Leitura dos dados brutos

# 1. Renda per capita estadual (Excel)
df_renda_per_capita = pd.read_excel(
    os.path.join(PASTA_DADOS, "Renda_per_capita_municipal.xlsx"),
    sheet_name="Tabela",
    skiprows=3
)

# 2. Empreendimentos de geracao distribuida - ANEEL (CSV)
df_aneel = pd.read_csv(
    os.path.join(PASTA_DADOS, "empreendimento-geracao-distribuida.csv"),
    sep=";",
    on_bad_lines="skip",
    encoding="ISO-8859-1",
    low_memory=False
)

# 3. Populacao, area e densidade demografica (Excel, 3 abas)
df_pop = pd.read_excel(
    os.path.join(PASTA_DADOS, "tabela4714_densidade_demografica.xlsx"),
    sheet_name="População residente (Pessoas)"
)
df_area = pd.read_excel(
    os.path.join(PASTA_DADOS, "tabela4714_densidade_demografica.xlsx"),
    sheet_name="Área da unidade territorial ..."
)
df_dens = pd.read_excel(
    os.path.join(PASTA_DADOS, "tabela4714_densidade_demografica.xlsx"),
    sheet_name="Densidade demográfica (Habit..."
)

# 4. Irradiacao solar GHI por MUNICIPIO - LABREN/INPE (CSV)
# Fonte: Atlas Brasileiro de Energia Solar, 2 ed. (Pereira et al., 2017)
# Traz o GHI medio anual e mensal da sede de cada um dos 5.569 municipios
# brasileiros. Substitui o ghi_stats.xlsx, que continha apenas estatisticas
# agregadas do pais e obrigava o uso de um valor unico por estado.
# Unidade original: Wh/m2/dia (convertido para kWh/m2/dia no Transform)
df_solar = pd.read_csv(
    os.path.join(PASTA_DADOS, "global_horizontal_means_sedes-munic.csv"),
    sep=";"
)
```

```python
# Persiste a Camada Bronze em disco para a proxima etapa (Transform.ipynb) ler
# Usamos pickle aqui (e nao parquet) porque as abas ainda tem cabecalhos/notas
# misturados nas colunas, o que so e limpo na etapa de Transform
df_renda_per_capita.to_pickle(os.path.join(PASTA_BRONZE, "renda_per_capita.pkl"))
df_aneel.to_pickle(os.path.join(PASTA_BRONZE, "aneel.pkl"))
df_pop.to_pickle(os.path.join(PASTA_BRONZE, "populacao.pkl"))
df_area.to_pickle(os.path.join(PASTA_BRONZE, "area.pkl"))
df_dens.to_pickle(os.path.join(PASTA_BRONZE, "densidade.pkl"))
df_solar.to_pickle(os.path.join(PASTA_BRONZE, "solar.pkl"))

print(f"Camada Bronze salva em: {os.path.abspath(PASTA_BRONZE)}")
```
